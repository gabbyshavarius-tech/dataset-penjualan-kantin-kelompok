```
import pandas as pd

df = pd.read_csv('/dataset_penjualan_kantin - dataset_penjualan_kantin.csv')

print(df.duplicated().sum()) # jumlah baris duplikat

df = df.drop_duplicates()

# Membersihkan harga_satuan
df['harga_satuan'] = df['harga_satuan'].fillna(0)
df['harga_satuan'] = df['harga_satuan'].astype(str)
df['harga_satuan'] = df['harga_satuan'].str.replace('Rp', '', regex=False)
df['harga_satuan'] = df['harga_satuan'].str.replace('.', '', regex=False)
df['harga_satuan'] = pd.to_numeric(df['harga_satuan'], errors='coerce').fillna(0).astype(int)

# Membersihkan jumlah_terjual
df['jumlah_terjual'] = df['jumlah_terjual'].fillna(0)
df['jumlah_terjual'] = df['jumlah_terjual'].astype(str)
df['jumlah_terjual'] = df['jumlah_terjual'].str.replace(' pcs', '', regex=False)
df['jumlah_terjual'] = pd.to_numeric(df['jumlah_terjual'], errors='coerce').fillna(0).astype(int)

print(df.dtypes)
```

```
laris = df[df['jumlah_terjual'] > 20] # filtering

urut = df.sort_values(by='jumlah_terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga_satuan'] * df['jumlah_terjual'] # kolom turunan

ringkasan = df.groupby('nama_produk')['total_pendapatan'].sum() # agregasi

print(ringkasan)
```
